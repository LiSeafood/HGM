```python
from HGM import compare_contrast_schemes
import pandas as pd

path = r"D:/study/project/Data/1.DLPFC/151507"

# 仅保留 gene_anchor，对比两种图像编码器：Phikon vs ViT
model_names = ["vit"]
all_detail = []
all_summary = []

for img_model_name in model_names:
    print(f"\n===== Running image encoder: {img_model_name} =====")
    detail_df, summary_df = compare_contrast_schemes(
        path=path,
        epochs=100,
        repeats=3,
        seed=2020,
        mode_list=["gene_anchor"],
        img_model_name=img_model_name,
    )

    detail_df["img_model_name"] = img_model_name
    summary_flat = summary_df.copy()
    summary_flat["img_model_name"] = img_model_name
    all_detail.append(detail_df)
    all_summary.append(summary_flat)

detail_all = pd.concat(all_detail, ignore_index=True)
summary_all = pd.concat(all_summary)

print("detail:")
print(detail_all)
print("\nsummary:")
print(summary_all)

# 可选：输出一个更直观的对比表（按编码器聚合均值）
summary_comp = summary_all.reset_index()
metric_cols = [
    c
    for c in summary_comp.columns
    if c not in ["img_model_name", "mode"] and pd.api.types.is_numeric_dtype(summary_comp[c])
]

if metric_cols:
    compare_table = summary_comp.groupby("img_model_name")[metric_cols].mean().sort_index()
    print("\ncompare_table (mean over repeats):")
    print(compare_table)
```
